## 5. TASK sql/python


### 1.Customer Analysis


##### How many customers are there by job type?

```python
df["job"].value_counts()
```

### 2. Loan Analysis

##### How many customers have loans?

```python
df["loan"].value_counts()
```

### 3. Deposit Subscription Analysis

##### How many customers subscribed for deposit?


```python
df["deposit"].value_counts()
```

### 4.Age Distribution


##### What is the age distribution of customers?

```python
df["age"].plot(kind="hist")
plt.show()
```

### 5. Average Age

```python
df["age"].mean()
```

### 6. Correlation Analysis

##### Which numeric columns are related?

```python
df.corr(numeric_only=True)
```

### 7. Correlation Heatmap

```python
import seaborn as sns

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.show()
```

### 8. Customers with Highest Balance

```python
df.sort_values(by="balance", ascending =False).head(10)
```

### 9. Average Balance by Job

```python
df.groupby("job")["balance"].mean()
```

### 10. Deposit Subscription Rate

```python
df["deposit"].value_counts(normalize=True) * 100
```

 ## Visualization

```python
df["job"].value_counts().plot(kind="bar")

plt.show()
```

```python
df["loan"].value_counts().plot( kind="pie",
    autopct="%1.1f%%"
)

plt.show()
```
